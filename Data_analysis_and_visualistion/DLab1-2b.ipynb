##Lab 1 – Part 2: Go Deeper (Tasks 1–6)

```python
import pandas as pd

listings = pd.read_csv("data/listings.csv")
```

##Task 1: Central or not

```python

# Task 1: Compare central and non-central listings

central = [
    'Södermalms',
    'Östermalms',
    'Norrmalms',
    'Kungsholmens'
]

# Identify listings in central neighbourhoods
mask = listings['neighbourhood_cleansed'].isin(central)

print("Central listings:", mask.sum())
print("Central share:", mask.mean())

# Compare private rooms
central_private = (
    listings[mask]['room_type'] == 'Private room'
).mean()

outside_private = (
    listings[~mask]['room_type'] == 'Private room'
).mean()

print("Private rooms in centre:", central_private)
print("Private rooms outside centre:", outside_private)
```

##Expected results from the worksheet:
Measure	Result
Central listings	2,331
Central share	51.0%
Private-room share in centre	15.1%
Private-room share outside	19.4%

Conclusion: Private rooms are proportionally more common outside central Stockholm.

isin() checks whether a value appears in your list. The ~ operator reverses a Boolean mask.

## Task 2: Room for a group

```python

# Task 2: Compare capacity for entire homes and other listings

entire = listings['room_type'] == 'Entire home/apt'

# Entire homes accommodating 4 or more guests
entire_share = (
    listings[entire]['accommodates'] >= 4
).mean()

# All other listing types
other_share = (
    listings[~entire]['accommodates'] >= 4
).mean()

print("Entire homes:", round(entire_share, 3))
print("Other listings:", round(other_share, 3))
```

##Task 3: Best reviewed, fairly

```python

# Task 3: Compare ratings for listings with at least 50 reviews

reviewed = listings[listings['number_of_reviews'] >= 50]

perfect = reviewed[reviewed['review_scores_rating'] == 5.0]

print("Perfect 5.0 ratings:", len(perfect))

columns = [
    'name',
    'neighbourhood_cleansed',
    'number_of_reviews',
    'review_scores_rating'
]

# Method 1: Sort by rating, then by number of reviews
top_sorted = reviewed.sort_values(
    ['review_scores_rating', 'number_of_reviews'],
    ascending=[False, False]
)[columns].head(5)

print("Method 1:")
print(top_sorted)

# Method 2: Select top five by rating only
top_nlargest = reviewed.nlargest(
    5, 'review_scores_rating'
)[columns]

print("Method 2:")
print(top_nlargest)
```

##Task 4: Label or position

```python

# Task 4: Understand loc, iloc, and reset_index

top = top_sorted.copy()

# Select by index label
print("Using loc:", top.loc[347, 'name'])

# Select by row position
print("Using iloc:", top.iloc[3]['name'])

# Reset the index to 0, 1, 2, 3, 4
top = top.reset_index(drop=True)

print(top)

# Intentionally produces a KeyError
#top.loc[347, 'name']
```

##Task 5: What does a zero mean?

```python

# Task 5A: Investigate listings with zero bedrooms

zero_bedrooms = listings[listings['bedrooms'] == 0]

print("Listings with 0 bedrooms:", len(zero_bedrooms))

print(zero_bedrooms['room_type'].value_counts())
```

##Part B: Zero availability

```python

# Task 5B: Investigate listings with no availability

zero_available = listings[listings['availability_365'] == 0]

print("Listings with zero availability:", len(zero_available))

no_reviews = (
    zero_available['number_of_reviews'] == 0
).sum()

print("Zero availability AND zero reviews:", no_reviews)
```
