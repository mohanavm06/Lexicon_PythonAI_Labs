```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. Set theme and load data
sns.set_theme(style='whitegrid')

date_columns = ['last_scraped', 'calendar_last_scraped', 'first_review', 'last_review']
listings = pd.read_csv('data/listings_clean.csv', parse_dates=date_columns)

# 2. Create a figure with two subplots to compare side-by-side
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Pure Matplotlib histogram
axes[0].hist(listings['number_of_reviews'].dropna(), bins=50, color='steelblue', edgecolor='black')
axes[0].set_title('Matplotlib: Number of Reviews Histogram')
axes[0].set_xlabel('Number of Reviews')
axes[0].set_ylabel('Frequency')

# Seaborn histogram
sns.histplot(data=listings, x='number_of_reviews', bins=50, ax=axes[1], color='teal')
axes[1].set_title('Seaborn: Number of Reviews Histplot')
axes[1].set_xlabel('Number of Reviews')
axes[1].set_ylabel('Count')

plt.tight_layout()
plt.show()
```
