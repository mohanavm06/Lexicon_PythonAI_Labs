```python

# Task 6: Inspect prices before sorting

print(listings['price'].head(3))
print("Data type:", listings['price'].dtype)

listings[['name', 'price']].sort_values(
    'price', ascending=False
).head(3)
```

```python

# Task 7A: Find listings with at least two bedrooms

mask = listings['bedrooms'] >= 2

print("Number:", mask.sum())
print("Share:", mask.mean())
```

```python

# Task 7B: Filter Östermalms entire homes with 2+ bedrooms

mask = (
    (listings['neighbourhood_cleansed'] == 'Östermalms')
    & (listings['room_type'] == 'Entire home/apt')
    & (listings['bedrooms'] >= 2)
)

selected = listings[mask]

print(selected.shape)
```

```python

# Task 8: Find the five most-reviewed listings in Östermalms

ostermalm = listings[
    listings['neighbourhood_cleansed'] == 'Östermalms'
]

top_five = ostermalm[
    ['name', 'room_type', 'number_of_reviews']
].sort_values(
    'number_of_reviews',
    ascending=False
).head(5)

print(top_five)
```
